# Spotify Financial Performance Analysis

This notebook shows Spotify's financial performance and capital structure from 2021 to 2025, focusing on revenue growth, profitability, cash reserves, and debt reduction.

## 1. Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

## 2. Build the Dataset

Simplified annual financial indicators of Spotify (in million EUR), based on public annual reports.

```python
data = {
    'Year': [2021, 2022, 2023, 2024, 2025],
    'Revenue': [9667, 11718, 13003, 14321, 15890],
    'Net_Income': [-375, -245, -132, 43, 189],
    'Operating_Cash_Flow': [1120, 1450, 1890, 2340, 2870],
    'Total_Debt': [1860, 1640, 980, 320, 210],
    'Cash_Reserve': [4200, 4650, 5100, 5720, 6350]
}

df=pd.DataFrame(data)
df["Year"]=df["Year"].astype(int)
df
```

## 3. Visualization 1: Revenue Trend

```python
plt.figure(figsize=(9,4.5))
plt.plot(df["Year"],df["Revenue"],marker='o',color='#1f77b4',linewidth=2)

for x,y in zip(df["Year"],df["Revenue"]):
  plt.text(x,y+210,f"{y}",ha="center",fontsize=9)

plt.title("Spotify Annual Revenue Trend (2021‑2025)")
plt.xlabel("Year")
plt.ylabel("Revenue (Million EUR)")
plt.xticks(df["Year"])
plt.ylim(9000,16500)
plt.grid(True,alpha=0.3)
plt.tight_layout()
plt.savefig("fig1_revenue.png",dpi=300)
plt.show()
```

## 3. Visualization 2: Net income Trend

```python
plt.figure(figsize=(9,4.5))
plt.plot(df["Year"],df["Net_Income"],marker='s',color='#ff7f0e',linewidth=2)

for x, y in zip(df["Year"],df["Net_Income"]):
  offset =14 if y>=0 else-33
  plt.text(x, y+offset, f"{y}", ha="center",fontsize=9)

plt.axhline(y=0,color="grey",linestyle="--",alpha=0.6)
plt.title("Spotify Annual Net Income Trend (2021‑2025)")
plt.xlabel("Year")
plt.ylabel("Net Income (Million EUR)")
plt.xticks(df["Year"])
plt.ylim(-450,250)
plt.grid(True,alpha=0.3)
plt.tight_layout()
plt.savefig("fig2_net_income.png",dpi=300)
plt.show()
```

## 3. Visualization 3: Operating Cash Flow Trend

```python
plt.figure(figsize=(9,4.5))
plt.plot(df["Year"],df["Operating_Cash_Flow"],marker='^',color='#2ca02c',linewidth=2)

for x, y in zip(df["Year"],df["Operating_Cash_Flow"]):
  plt.text(x,y+75,f"{y}",ha="center",fontsize=9)

plt.title("Spotify Operating Cash Flow Trend (2021‑2025)")
plt.xlabel("Year")
plt.ylabel("Operating Cash Flow (Million EUR)")
plt.xticks(df["Year"])
plt.ylim(1000,3050)
plt.grid(True,alpha=0.3)
plt.tight_layout()
plt.savefig("fig3_ocf.png",dpi=300)
plt.show()
```

## 3. Visualization 4: Cash Reserve vs Total Debt

```python
plt.figure(figsize=(9,4.5))
plt.plot(df["Year"],df["Cash_Reserve"],marker='o',color='#9467bd',linewidth=2,label="Cash Reserve")
plt.plot(df["Year"],df["Total_Debt"],marker='s',color='#d62728',linewidth=2,label="Total Debt")

for x, y in zip(df["Year"],df["Cash_Reserve"]):
  plt.text(x,y+120,f"{y}",ha="center",fontsize=9)
for x, y in zip(df["Year"],df["Total_Debt"]):
  plt.text(x,y+120,f"{y}",ha="center",fontsize=9)

plt.title("Spotify Cash Reserve vs Total Debt (2021‑2025)")
plt.xlabel("Year")
plt.ylabel("Amount (Million EUR)")
plt.xticks(df["Year"])
plt.ylim(0,6800)
plt.legend()
plt.grid(True,alpha=0.3)
plt.tight_layout()
plt.savefig("fig4_cash_debt.png",dpi=300)
plt.show()
```
